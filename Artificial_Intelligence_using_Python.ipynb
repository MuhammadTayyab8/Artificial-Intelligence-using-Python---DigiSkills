```python
# OOP

# CLASS + OBJECT
class Student:
  def say_hello(self):
    print("Hello Student!")

student1 = Student()
student1.say_hello()

print("\n------------")

# Constructor (init) and Instance Methods
class Student:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bio(self):
    print(f"Name: {self.name}, Age: {self.age}")


student1 = Student("Muhammad Tayyab", 18)
print(student1.name)
print(student1.age)
student1.bio()

print("\n------------")

# Encapsulation (Public/Private)
class BankAccount:
  def __init__(self):
    self.__balance = 1000

account = BankAccount()
# This will give error
# print(account.__balance)

print("\n------------")

# Getters & Setters
class BankAccount:
  def __init__(self):
    self.__balance = 1000

  def get_balance(self):
    return self.__balance

  def set_balance(self, new_balance):
    self.__balance = new_balance

account = BankAccount()
print(account.get_balance())
account.set_balance(1500)
print(account.get_balance())

print("\n------------")

# Inheritance
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

dog1 = Dog()

dog1.speak()
dog1.bark()

print("\n------------")

# METHOD OVERRIDING
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

dog1 = Dog()

dog1.speak()

print("\n------------")

# POLYMORPH
class Bird:

    def sound(self):
        print("Bird sound")


class Parrot(Bird):

    def sound(self):
        print("Parrot talking")


class Crow(Bird):

    def sound(self):
        print("Crow cawing")


birds = [Parrot(), Crow()]

for bird in birds:
    bird.sound()


# SPECIAL METHODS
class Book:

    def __init__(self, title):
        self.title = title

    def __str__(self):
        return f"Book: {self.title}"


book1 = Book("Python Basics")

print(book1)
```

```python
# Assignment 1 — Calculator Class
# Create class:
# Calculator
# Methods:
# add
# subtract
# multiply
# divide

class Calculator:
  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2

  def add(self):
    print(self.num1 + self.num2)

  def subtract(self):
    print(self.num1 - self.num2)

  def multiply(self):
    print(self.num1 * self.num2)

  def divide(self):
    print(self.num1 / self.num2)

calculator = Calculator(10, 5)
calculator.add()
calculator.subtract()
calculator.multiply()
calculator.divide()


# Assignment 2 — Student Class
# Attributes:
# name
# age
# course
# Method:
# display_info()

print("\n=============")

class Student:
  def __init__(self, name, age, course):
    self.name = name
    self.age = age
    self.course = course

  def display_info(self):
    print(f"Name: {self.name}, Age: {self.age}, Course: {self.course}")

student = Student("Muhammad Tayyab", 18, "Artificial Intelligence")
student.display_info()

print("\n=============")


# Assignment 6 ⭐ MINI PROJECT
# Create simple:
# Bank Account System
# Features:
# deposit
# withdraw
# show balance

class Bank:
  def __init__(self):
    self.__balance = 0

  def deposit(self, amount):
    self.__balance += amount
    print(f"Deposit: {amount}")

  def withdraw(self, amount):
    if amount > self.__balance:
      print("Insufficient balance")
      return
    self.__balance -= amount
    print(f"Withdraw: {amount}")

  def show_balance(self):
    print(f"Balance: {self.__balance}")

account = Bank()
account.deposit(1000)
account.withdraw(600)
account.show_balance()
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={
        "Category": "category",
        "Region": "category"
    }
)

df.head()

df.info()

# Handling Missing Data
print(df.isna().sum())  # every column null values sum


df["Quantity"] = pd.to_numeric(df["Quantity"])
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mode()[0])

df["Sales"] = df["Sales"].fillna(df["Sales"].median())

df["Category"] = df["Category"].fillna(df["Category"].mode()[0])


print("\n ===== After fillna")
df.info()

# Handling Missing Data
print(df.isna().sum())  # every column null values sum


# drop region
df.dropna(subset=["Region"], inplace=True)

print("\n ===== After dropna")
df.info()
# Handling Missing Data
print(df.isna().sum())  # every column null values sum
```
